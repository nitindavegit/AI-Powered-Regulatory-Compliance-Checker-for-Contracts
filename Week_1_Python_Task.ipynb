# ***2. Strings & String Methods***



````python
s = 'Nitin Dave'

# Common String Methods
print(s.upper())
print(s.lower())
print(s.split())
print(s.replace("Dave","Ghadkari"))
print(s.strip())
print(s.startswith('Nitin'))
print(s.find('ave'))
print(len(s))

print("\n\n\n")

# String Slicing
print(s[1:5])
print(s[::-1])

# f-String Formatting
name = "Nitin"
age = 20
print(f"Name: {name}, Age: {age}")

print("\n\n\n")

# String is iterable
for char in 'nitin':
  print(char, end=' ')

print("\n\n\n")
# Interview Q: Check if a string is palindrome
def pallindrome(str):
  formatted_str = str.strip().lower()
  reverse_str = formatted_str[::-1]
  if reverse_str == formatted_str:
    return "Pallindrome"
  return "Not a Pallindrome"

print(pallindrome('Nitin'))
print(pallindrome("Hello"))
````

# ***5. Functions, ```*args```, ```**kwargs```***





````python
# Basic function

def greet(name, greeting='Hello'):
  return f"{greeting}, {name}"

print(greet("Nitin","Hey"))
print(greet("Jay"))

# *args
def sum_all(*args):
  return sum(args)

print(sum_all(1,2,3,4,5,6))

# **kwargs
def show_info(**kwargs):
  for key, value in kwargs.items():
    print(f"key: {key}, value: {value}")

show_info(name='Nitin', age=20, city='Surat')
````

# ***8. OOP (Object Oriented Programming)***

````python
# class and object

class Animal:
  species = 'Animal'

  def __init__(self, name, age):
    self.name = name
    self.age = age

  def speak(self):
    return f'{self.name} makes a sound'

  def __str__(self):
    return f'Animal({self.name}, {self.age})'

  def __repr__(self):
    return f'Animal(name={self.name!r}, age={self.age!r})'


a = Animal('Generic',3)
print(a.__str__())
print(a.__repr__())


# Inheritance
class Dog(Animal):
  def __init__(self, name, age, breed):
    super().__init__(name, age)
    self.breed = breed

  def speak(self):     # Method Overriding
    return f"{self.name} says Wooo Wooo"

class Cat(Animal):
  def speak(self):     # Method Overriding
    return f"{self.name} says Meow Meow!"


dog = Dog('Bruno',5,'Pomerian')
cat = Cat('whiskiers',3)

print(dog.speak())
print(cat.speak())


# Polymorphism
animals = [dog, cat, Animal('Bird', 2)]
for animal in animals:
    print(animal.speak())  # Same method, different behavior


# Encapsulation
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())  # 1500


# Abstraction
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return 3.14159 * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, w, h):
        self.w, self.h = w, h
    def area(self):
        return self.w * self.h

for shape in [Circle(5), Rectangle(4, 6)]:
    print(f'Area: {shape.area()}')
````
